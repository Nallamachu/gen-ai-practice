#### Introduction to NLP

Natural Language Processing (NLP) combines linguistics and machine learning to process human language (text or speech). In deep learning, NLP uses neural networks to automatically learn patterns from large text datasets instead of relying on hand-crafted rules.

#### Why is NLP important?

It powers many everyday applications:
- Chatbots (Like ChatGPT)
- Translation (Google Translate)
- Voice Assist (Siri, Alexa)
- Sentiment Analysis
- Search Engines

#### Traditional VS DeepLearning NLP
##### Traditional NLP
- Rule Based or Statistical Methods
- Heavy feature engineering (Manual Effort)
- Ngram, TF-IDF

##### Deep Learning NLP
- Uses neural network
- Automatically learn features from data
- Handles complex patterns like context and meaning

#### Key concepts in Deeplearning NLP
1. Text Representation
Machines don't understand words - they understand numbers

##### a. one hot encoding
- Each word -> Vector of 0s with 1s
- Problem: sparse and no meaning captured

##### b. Word Embedding
Dense vectors and captures meaning

Popular Methods:
- Word2vec
- GloVe
- FastText

#### Neural Network Architecture for NLP
##### a. RNN - Recurrent Neural Network
- Designed for sequence data
- Remembers Previous Input

##### b. LSTM - Long Short Term Memory
- Improved RNN
- Solves vanishing gradient problem
- Handles long-term dependencies

##### c. GRU - Gated Recurrent Unit
- Simpler version of LSTM
- Faster Training

##### d. Transformers
Based on the concept of attention mechanism, focuses on important words in a sentense regardless of position.

Popular Models:
- BERT
- GPT
- T5


#### NLP Pipiline
1. Text Collection
2. Preprocessing
    - Lowercasing
    - Removing Punctuation
    - Stopward removal
    - Stemming / Lemmatization
3. Tokenization
    - Breaking text into words and subwords
4. Vectorization
    - Convert text -> embeddings
5. Model Training
    - Use neural network
6. Evaluation
    - Accuracy, F1-score, BLEU

#### Common NLP Tasks
1. Text Classification
    - Spam Detection
    - Sentiment Analysis
2. Named Entity Recognition - NER
    - Identify name, places, dates
3. Machine Translation
    - English -> Telugu, Hindi, etc
4. Text Generation
    - Chatbots, story generation
5. Question Answering
    - Like ChatGPT
