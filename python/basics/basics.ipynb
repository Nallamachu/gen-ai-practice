## Indentation

```python
age= 25

if age >= 18:
    print("You are eligible to vote.")
else:
    print("You are not eligible to vote.")
```

## Semantics in Python : kernel automatically identify the type of the variable based on the value assigned

```python
age = 30    ## age is an integer variable
name = "Subbareddy" ## name is a string variable

print(type(age))  ## prints <class 'int'>
print(type(name))  ## prints <class 'str'>
```

## Type Inference
Runtime a variable can be changed to different types

```python
variable=10
print(type(variable))  ## prints <class 'int'>
variable="Subbareddy"
print(type(variable))  ## prints <class 'str'>
```
